## Actividad 3 — Primer `pipeline`

Carga un `pipeline` ligero (por ejemplo sentimiento **en inglés**) y evalúa **dos frases** que tú elijas: una claramente positiva y otra ambigua.

*Hint:* `from transformers import pipeline` — if memory is tight, pick a small model (e.g. a DistilBERT-sized checkpoint).


```python
# from transformers import pipeline
# clf = pipeline(...)
# TODO: run on two strings and print scores
from transformers import pipeline
clf = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

frases = [
    "I absolutely love this new framework, it is blazing fast!",
    "The movie was okay, not great but not terrible either."
]

resultados = clf(frases)
for f, r in zip(frases, resultados):
    print(f"Frase: '{f}' -> {r}")
```
