## Actividad 3 — Auto-crítica

Diseña un prompt en dos pasos: (1) el modelo responde, (2) el mismo modelo evalúa su respuesta con una checklist (alucinación, completitud). Implementa como **dos strings** que puedas encadenar en tu laboratorio.


````python
step1 = """
Eres un catedrático experto en Inteligencia Artificial. Explica de forma detallada qué es el mecanismo de atención (Attention Mechanism) en los modelos Transformer y cómo resuelve las limitaciones de las redes recurrentes (RNN).
"""

step2 = """
Revisa la explicación técnica generada en el paso anterior y actúa como un revisor crítico de artículos científicos. Evalúa el texto utilizando la siguiente lista de verificación (checklist):

1. ¿Contiene alucinaciones técnicas o imprecisiones matemáticas sobre las matrices Query, Key y Value?
2. ¿Es una respuesta completa o ha omitido explicar detalladamente el problema del desvanecimiento del gradiente en secuencias largas de las RNN?
3. ¿El tono es riguroso, formal y estrictamente académico?

Produce un informe final estructurado de la siguiente manera:
- [AUDITORÍA]: Breve evaluación de los puntos anteriores señalando posibles fallos o carencias encontradas en el texto original.
- [RESPUESTA REFINADA]: El texto corregido y mejorado, subsanando cualquier debilidad detectada en la auditoría para garantizar la máxima completitud y veracidad.
"""

print("=== PROMPT: PASO 1 (GENERACIÓN) ===")
print(step1.strip())
print("\n=== PROMPT: PASO 2 (AUTO-CRÍTICA Y REFINAMIENTO) ===")
print(step2.strip())
````
