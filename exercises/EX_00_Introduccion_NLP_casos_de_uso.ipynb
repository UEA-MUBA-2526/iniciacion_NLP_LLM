## Actividad 3 — Primer `pipeline`

Carga un `pipeline` ligero (por ejemplo sentimiento **en inglés**) y evalúa **dos frases** que tú elijas: una claramente positiva y otra ambigua.

*Hint:* `from transformers import pipeline` — if memory is tight, pick a small model (e.g. a DistilBERT-sized checkpoint).


```python
from transformers import pipeline

# 1. Cargamos el pipeline de análisis de sentimiento con el modelo ligero que analizamos antes
# Al usar este checkpoint ligero (DistilBERT), la descarga será muy rápida y consumirá poca memoria.
clf = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

# 2. Definimos las dos frases en inglés (una muy positiva y otra con trampa/ambigua)
frase_positiva = "This machine learning framework is absolutely fantastic and makes my life so much easier!"
frase_ambigua  = "The movie was not terrible, but I wouldn't say it was a masterpiece either."

# 3. Evaluamos las frases con el pipeline
resultado_positivo = clf(frase_positiva)[0]
resultado_ambiguo  = clf(frase_ambigua)[0]

# 4. Imprimimos los resultados con un formato limpio
print("--- RESULTADOS DEL PIPELINE ---")
print(f"Frase Positiva: '{frase_positiva}'")
print(f" -> Predicción: {resultado_positivo['label']} | Confianza (Score): {resultado_positivo['score']:.4f}\n")

print(f"Frase Ambigua:  '{frase_ambigua}'")
print(f" -> Predicción: {resultado_ambiguo['label']} | Confianza (Score): {resultado_ambiguo['score']:.4f}")
```
