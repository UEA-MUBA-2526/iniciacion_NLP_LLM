## Actividad 3 — Primer `pipeline`

Carga un `pipeline` ligero (por ejemplo sentimiento **en inglés**) y evalúa **dos frases** que tú elijas: una claramente positiva y otra ambigua.

*Hint:* `from transformers import pipeline` — if memory is tight, pick a small model (e.g. a DistilBERT-sized checkpoint).


```python
from transformers import pipeline

# Sentiment analysis pipeline
clf = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# Clearly positive sentence
positive_text = "I absolutely loved this course, it was extremely useful and well explained."

# Ambiguous sentence
ambiguous_text = "The movie was interesting, although some parts felt too long."

# Predictions
positive_result = clf(positive_text)
ambiguous_result = clf(ambiguous_text)

print("Positive sentence result:")
print(positive_result)

print("\nAmbiguous sentence result:")
print(ambiguous_result)
```
