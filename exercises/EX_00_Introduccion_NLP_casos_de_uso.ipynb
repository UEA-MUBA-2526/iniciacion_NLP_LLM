## Actividad 3 — Primer `pipeline`

Carga un `pipeline` ligero (por ejemplo sentimiento **en inglés**) y evalúa **dos frases** que tú elijas: una claramente positiva y otra ambigua.

*Hint:* `from transformers import pipeline` — if memory is tight, pick a small model (e.g. a DistilBERT-sized checkpoint).


```python
from transformers import pipeline

# 1. Cargamos el pipeline ligero de análisis de sentimiento (sentiment-analysis)
# Por defecto descarga un modelo rápido basado en DistilBERT optimizado para inglés
clf = pipeline("sentiment-analysis")

# 2. Definimos las dos frases solicitadas
# Frase 1: Claramente positiva
positive_sentence = "This new deep learning framework is absolutely amazing and easy to use!"

# Frase 2: Ambigua (tiene cosas buenas y malas, ideal para desafiar al modelo)
ambiguous_sentence = "The movie had stunning visuals, but the plot was completely boring."


# =====================================================================
# TODO: run on two strings and print scores
# =====================================================================

# Evaluamos las frases pasándoselas al pipeline
result_positive = clf(positive_sentence)[0]
result_ambiguous = clf(ambiguous_sentence)[0]

# --- Mostrar los resultados en pantalla ---
print("--- EVALUACIÓN DE FRASES ---")
print(f"\nFrase Positiva: '{positive_sentence}'")
print(f"  -> Predicción: {result_positive['label']} | Confianza: {result_positive['score']:.4f}")

print(f"\nFrase Ambigua: '{ambiguous_sentence}'")
print(f"  -> Predicción: {result_ambiguous['label']} | Confianza: {result_ambiguous['score']:.4f}")
```
