## Actividad 3 — Primer `pipeline`

Carga un `pipeline` ligero (por ejemplo sentimiento **en inglés**) y evalúa **dos frases** que tú elijas: una claramente positiva y otra ambigua.

*Hint:* `from transformers import pipeline` — if memory is tight, pick a small model (e.g. a DistilBERT-sized checkpoint).


```python
# from transformers import pipeline
# clf = pipeline(...)
# TODO: run on two strings and print scores
from transformers import pipeline

# Pipeline de análisis de sentimiento en español
sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment",
    device=-1)

frases= [
    "Me encanta este producto, es increíble.",
    "No me gusta este producto, es terrible.",
    "Es un producto regular, no está mal pero tampoco es excelente."
]

for frase in frases:
    resultado = sentiment_analyzer(frase)
    print(f"Frase: '{frase}' -> Sentimiento: {resultado[0]['label']} (score: {resultado[0]['score']:.4f})")
```
