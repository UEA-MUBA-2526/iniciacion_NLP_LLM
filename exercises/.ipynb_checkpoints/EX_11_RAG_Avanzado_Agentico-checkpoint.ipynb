## Actividad 1 — Self-RAG checklist

Lista 3 preguntas binarias que el sistema podría autoevaluar sobre un borrador (soporte en contexto, completitud, seguridad). Respuesta en markdown.


1. **¿Soporte en contexto (Factualidad/Faithfulness)?** -> ¿Toda la información contenida en la respuesta generada se deriva estrictamente del contexto recuperado, asegurando que no se ha inventado información externa ni se sufre de alucinaciones? (SÍ / NO)
2. **¿Completitud (Answer Relevance)?** -> ¿La respuesta responde de forma íntegra a todas las intenciones, dudas o sub-preguntas planteadas por el usuario en su consulta original sin omitir información clave? (SÍ / NO)
3. **¿Seguridad y Utilidad (Safety/Harmful Content)?** -> ¿La respuesta mantiene un tono ético, profesional, y está completamente libre de sesgos, alucinaciones peligrosas, contradicciones o información dañina? (SÍ / NO)
