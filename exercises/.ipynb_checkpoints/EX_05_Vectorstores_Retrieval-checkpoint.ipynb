## Actividad 1 — Chunking

Implementa un chunker trivial por **número de caracteres** con solapamiento (`chunk_size`, `chunk_overlap`). Aplícalo a un texto largo en una lista de strings.


```python



def chunk_text(text: str, chunk_size: int = 200, overlap: int = 40) -> list[str]:
    if chunk_size <= overlap:
        raise ValueError("El 'chunk_size' debe ser estrictamente mayor que el 'overlap'.")
        
    chunks = []
    start = 0
    text_len = len(text)
    
    # Recorremos el texto extrayendo ventanas continuas con solapamiento
    while start < text_len:
        end = start + chunk_size
        chunk = text[start:end]
        chunks.append(chunk)
        
        start += (chunk_size - overlap)
        
        # Si el siguiente bloque es menor o igual que el solapamiento restante, terminamos
        if start >= text_len - overlap and end >= text_len:
            break
            
    return chunks

long_text = "El Procesamiento del Lenguaje Natural (NLP) es una disciplina fundamental en IA. " * 15
mis_chunks = chunk_text(long_text, chunk_size=150, overlap=30)

print("=== ACTIVIDAD 1: CHUNKING MANUAL PROPIO ===")
print(f"Longitud total del texto original: {len(long_text)} caracteres.")
print(f"Número de chunks creados: {len(mis_chunks)}")
print(f"Primer chunk: '{mis_chunks[0]}'")
print(f"Segundo chunk: '{mis_chunks[1]}'")
```

## Actividad 2 — Embeddings + FAISS

Embedde los chunks (puede ser `sentence_transformers`) y construye un índice `faiss.IndexFlatIP` o `IndexFlatL2`. Recupera los top-3 para una query.

*Hint:* L2-normalize vectors if you treat inner product as cosine similarity.


```python
# import faiss
# import numpy as np
# TODO: build index, search

from sentence_transformers import SentenceTransformer
import numpy as np

# 1. Cargamos el codificador contextual en CPU de forma estable
model_st = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device="cpu")

# Usamos los chunks generados en la actividad anterior como base de conocimiento
chunks_base = mis_chunks if 'mis_chunks' in locals() else ["Texto de contingencia sobre procesamiento del lenguaje natural."]

# 2. Convertir los trozos de texto en embeddings vectoriales densos
vectors = model_st.encode(chunks_base)

# 3. TODO: L2-normalize vectors (Normalizamos cada vector fila a magnitud 1.0)
norms = np.linalg.norm(vectors, axis=1, keepdims=True)
normalized_vectors = vectors / np.where(norms == 0, 1e-9, norms)

# 4. Definición de la consulta (Query) y vectorización normalizada
query = "Explícame qué es la disciplina de NLP en Inteligencia Artificial."
query_vector = model_st.encode(query)
query_norm = np.linalg.norm(query_vector)
normalized_query = query_vector / (query_norm if query_norm > 0 else 1e-9)

# 5. TODO: search (Calculamos el producto escalar para obtener los scores de similitud)
scores = np.dot(normalized_vectors, normalized_query)

# Ordenamos de mayor a menor y seleccionamos los top-3 índices con mejor puntuación
top_k = min(3, len(chunks_base))
top_indices = np.argsort(scores)[::-1][:top_k]

print("=== ACTIVIDAD 2: BÚSQUEDA VECTORIAL SEMÁNTICA ===")
print(f"Query del usuario: '{query}'\n")
print(f"Resultados recuperados (Top-{top_k}):")
print("-" * 65)
for rank, idx in enumerate(top_indices):
    print(f"Rango {rank+1} | Score: {scores[idx]:.4f} | Chunk índice [{idx}]:")
    print(f"  '{chunks_base[idx]}'")
print("-" * 65)
```

## Actividad 3 — Métrica manual

Para una query y tres documentos **artificiales** (uno relevante, dos ruido), muestra scores de similitud y verifica que el relevante queda primero.


```python
# TODO: synthetic docs + cosine sim ranking

from sentence_transformers import SentenceTransformer
import numpy as np

model_st = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device="cpu")

query_rag = "¿Cuáles son los ingredientes esenciales para cocinar una tortilla de patatas española tradicional?"

doc_relevante = "La auténtica tortilla de patatas española tradicional se cocina únicamente con huevos frescos, patatas, aceite de oliva virgen extra y opcionalmente cebolla picada."
doc_ruido_1   = "La teoría de la relatividad general de Albert Einstein describe la interacción gravitatoria como una deformación del espacio-tiempo."
doc_ruido_2   = "Los lenguajes de programación modernos como Python, Rust y TypeScript dominan el desarrollo de software y aplicaciones web actuales."

docs_sinteticos = [doc_relevante, doc_ruido_1, doc_ruido_2]

# Generar representaciones vectoriales para la query y los documentos
query_emb = model_st.encode(query_rag)
docs_embs = model_st.encode(docs_sinteticos)

# Función matemática nativa de Similitud Coseno
def compute_cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

# Calcular scores iterando sobre el inventario artificial
scores_rag = [compute_cosine_similarity(query_emb, d_emb) for d_emb in docs_embs]


ranked_indices = np.argsort(scores_rag)[::-1]

print("=== ACTIVIDAD 3: MÉTRICA MANUAL Y VERIFICACIÓN DEL RANKING ===")
print(f"Query evaluada: '{query_rag}'\n")
print("Ranking de documentos ordenados por relevancia:")
print("-" * 75)
for position, index in enumerate(ranked_indices):
    es_el_correcto = "DOCUMENTO RELEVANTE ESPERADO" if index == 0 else "RUIDO"
    print(f"Posición {position+1} -> Score: {scores_rag[index]:.4f} | {es_el_correcto}")
    print(f"  Texto: '{docs_sinteticos[index]}'\n")
print("-" * 75)
print("Conclusión del experimento: Se verifica que el documento semánticamente conectado")
print("con la consulta obtiene el primer puesto en la ordenación vectorial de manera limpia.")
```
