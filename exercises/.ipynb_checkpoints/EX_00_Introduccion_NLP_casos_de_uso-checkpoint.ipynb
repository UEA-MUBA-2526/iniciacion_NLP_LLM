## Actividad 3 — Primer `pipeline`

Carga un `pipeline` ligero (por ejemplo sentimiento **en inglés**) y evalúa **dos frases** que tú elijas: una claramente positiva y otra ambigua.

*Hint:* `from transformers import pipeline` — if memory is tight, pick a small model (e.g. a DistilBERT-sized checkpoint).


```python
from transformers import pipeline


clf = pipeline(
    "sentiment-analysis", 
    model="distilbert-base-uncased-finetuned-sst-2-english",
    device=-1
)

# Definimos las dos frases
frase_positiva = "This introductory NLP course is absolutely fantastic, I am learning a lot!"
frase_ambigua = "The movie was long and slow, but the ending made me rethink the whole story."

resultado_positivo = clf(frase_positiva)[0]
resultado_ambiguo = clf(frase_ambigua)[0]

# TODO: run on two strings and print scores
print("=== EVALUACIÓN DE FRASES ===")
print(f"Frase Positiva: '{frase_positiva}'")
print(f"Resultado: {resultado_positivo['label']} | Confianza: {resultado_positivo['score']:.4f}\n")

print(f"Frase Ambigua: '{frase_ambigua}'")
print(f"Resultado: {resultado_ambiguo['label']} | Confianza: {resultado_ambiguo['score']:.4f}")
```
