## Actividad 1 — Reordenar por cross-score simulado

Dada una query y 5 documentos, supón que tienes scores de un bi-encoder (baratos) y scores de un cross-encoder (caros). Implementa: tomar top-4 por bi-encoder y reordenar solo esos 4 por cross-score.


```python
# TODO: two-stage ranking, print final order

import numpy as np

query = "latency vs throughput"
docs = [f"doc{i}: content explaining network metrics" for i in range(5)]

bi_scores = np.array([0.72, 0.85, 0.41, 0.68, 0.91])
cross_scores_simulated = {
    0: 0.88,
    1: 0.65,
    3: 0.74,
    4: 0.95
}

print("=== ACTIVIDAD 1: RETRIEVAL EN DOS ETAPAS (TWO-STAGE RANKING) ===")

# Obtenemos los índices ordenados de mayor a menor según el Bi-Encoder
indices_ordenados_bi = np.argsort(bi_scores)[::-1]
top_4_bi_indices = indices_ordenados_bi[:4]

print(f"Paso 1 (Filtro Bi-Encoder) - Índices top-4 seleccionados: {list(top_4_bi_indices)}")

# Extraemos las puntuaciones del Cross-Encoder para los 4 elementos seleccionados
final_scores = []
for idx in top_4_bi_indices:
    score_ce = cross_scores_simulated[idx]
    final_scores.append((idx, score_ce))

# Ordenamos la lista de tuplas de mayor a menor basándonos en el score del Cross-Encoder
final_scores.sort(key=lambda x: x[1], reverse=True)

print("\nPaso 2 (Reordenamiento Cross-Encoder) - Ranking Final:")
print("-" * 65)
for rank, (idx, score) in enumerate(final_scores):
    print(f"Piso {rank+1} -> Índice Original Documento: [{idx}] | Cross-Score: {score:.2f} | {docs[idx]}")
print("-" * 65)
```

## Actividad 2 — MMR esquemático

En pseudocódigo en Python (sin librería), bosqueja 5 líneas de selección **MMR** (balance relevancia / diversidad).


```python
# TODO: MMR pseudocode as comments or stub function

def maximal_marginal_relevance(query_vector, candidate_vectors, lambda_param=0.5, top_k=3):
    """
    Bosquejo conceptual del algoritmo MMR para balancear Relevancia vs Diversidad.
    
    Parámetros:
      lambda_param: 1.0 busca pura relevancia; 0.0 busca pura diversidad de información.
    """
    selected_indices = []
    # Clonamos la lista de candidatos disponibles para ir purgando
    remaining_indices = list(range(len(candidate_vectors)))
    
    # Bucle principal de selección iterativa paso a paso
    for _ in range(top_k):
        if not remaining_indices:
            break
            
        best_mmr_score = -float('inf')
        best_candidate = None
        
        # Evaluamos la ecuación marginal sobre cada candidato restante
        for i in remaining_indices:
            relevance = cosine_similarity(candidate_vectors[i], query_vector)
            if len(selected_indices) == 0:
                redundancy = 0.0
            else:
                redundancy = max([cosine_similarity(candidate_vectors[i], candidate_vectors[s]) for s in selected_indices])
            
            # Ecuación fundamental del MMR:
            mmr_score = lambda_param * relevance - (1 - lambda_param) * redundancy
            
            # Guardamos el candidato con mejor beneficio marginal neto
            if mmr_score > best_mmr_score:
                best_mmr_score = mmr_score
                best_candidate = i
                
        # Movemos el ganador del almacén de candidatos al de seleccionados
        selected_indices.append(best_candidate)
        remaining_indices.remove(best_candidate)
        
    return selected_indices

print("=== ACTIVIDAD 2: ESQUEMA LÓGICO MMR CONFIGURADO ===")
print("El esqueleto del algoritmo MMR ha sido definido con éxito respetando la penalización de redundancia.")
```
