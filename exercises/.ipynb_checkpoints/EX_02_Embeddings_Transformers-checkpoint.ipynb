## Actividad 1 — Mean pooling

Con `AutoTokenizer` + `AutoModel`, obtén **last_hidden_state** para una frase y calcula el embedding de frase como media sobre tokens (excluyendo padding).


```python
from transformers import AutoTokenizer, AutoModel
import torch

model_name = "sentence-transformers/all-MiniLM-L6-v2"
text = "Transformers build contextual embeddings."

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name, use_safetensors=True)
model.eval()

#TODO: tokenize 
encoded_input = tokenizer(text, padding=True, truncation=True, return_tensors="pt")

# TODO: forward 
with torch.no_grad():
    model_output = model(**encoded_input)

# Extraemos el last_hidden_state
last_hidden_state = model_output.last_hidden_state
attention_mask = encoded_input["attention_mask"]

# 5. TODO: mean pool 
# Multiplicamos los embeddings por la máscara de atención para poner a 0 los tokens de padding
input_mask_expanded = attention_mask.unsqueeze(-1).expand(last_hidden_state.size()).float()
sum_embeddings = torch.sum(last_hidden_state * input_mask_expanded, 1)

# Calculamos el número real de tokens válidos por frase (evitando división por cero)
sum_mask = torch.clamp(input_mask_expanded.sum(1), min=1e-9)
sentence_embedding = (sum_embeddings / sum_mask).squeeze(0).numpy()

print("=== ACTIVIDAD 1: MEAN POOLING MANUAL ===")
print(f"Dimensiones del last_hidden_state: {last_hidden_state.shape}")
print(f"Dimensiones del embedding de frase final: {sentence_embedding.shape}")
print("\nPrimeros 5 valores del embedding generado:")
print(sentence_embedding[:5])
```

## Actividad 2 — `sentence-transformers`

Usa `SentenceTransformer` para embedder dos frases y calcula similitud coseno. Comenta brevemente (en inglés en un comentario) por qué suele ser mejor que mean-pooling manual de BERT base.


```python
from sentence_transformers import SentenceTransformer
import numpy as np

# Cargar el modelo especializado a través de SentenceTransformer
model_st = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device="cpu")

# Definir dos frases para comparar su similitud
sentence1 = "The weather is beautiful today in Madrid."
sentence2 = "It is a lovely sunny day outside."

embedding1 = model_st.encode(sentence1)
embedding2 = model_st.encode(sentence2)

#TODO: cosine similarity
def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

similarity_score = cosine_similarity(embedding1, embedding2)

print("=== ACTIVIDAD 2: SENTENCE TRANSFORMERS ===")
print(f"Frase 1: '{sentence1}'")
print(f"Frase 2: '{sentence2}'")
print(f"Similitud Coseno: {similarity_score:.4f}")
```

## Actividad 3 — Paráfrasis

Escribe dos paráfrasis de una misma idea y muestra que sus embeddings (sentence-transformers) tienen **mayor** similitud entre sí que con una frase de tema distinto.


```python
# TODO: three strings, 2 paraphrase + 1 unrelated, print cosine sims

from sentence_transformers import SentenceTransformer
import numpy as np

# Reutilizamos el modelo cargado en la actividad anterior
model_st = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device="cpu")

#TODO: three strings, 2 paraphrase + 1 unrelated
phrase_a = "Artificial Intelligence is transforming the global job market rapidly."
phrase_b = "The landscape of modern employment is being reshaped by AI technologies."
phrase_c = "Homemade chocolate cake requires high-quality cocoa powder and fresh butter."

#Extraer los embeddings correspondientes
emb_a = model_st.encode(phrase_a)
emb_b = model_st.encode(phrase_b)
emb_c = model_st.encode(phrase_c)

# Función auxiliar de similitud coseno
def compute_cos_sim(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

#Calcular los contrastes semánticos
sim_paraphrase = compute_cos_sim(emb_a, emb_b)  # Relación A con B
sim_unrelated_1 = compute_cos_sim(emb_a, emb_c) # Relación A con C
sim_unrelated_2 = compute_cos_sim(emb_b, emb_c) # Relación B con C

# TODO: print cosine sims
print("=== ACTIVIDAD 3: EVALUACIÓN DE PARÁFRASIS ===")
print(f"Frase A: '{phrase_a}'")
print(f"Frase B (Paráfrasis): '{phrase_b}'")
print(f"Frase C (No relacionada): '{phrase_c}'\n")

print("-" * 60)
print(f"Similitud entre Paráfrasis (A vs B):       {sim_paraphrase:.4f} (Alta)")
print(f"Similitud con No Relacionada (A vs C):    {sim_unrelated_1:.4f} (Baja)")
print(f"Similitud con No Relacionada (B vs C):    {sim_unrelated_2:.4f} (Baja)")
print("-" * 60)
print("\nConclusión: El espacio vectorial del modelo discrimina con precisión")
print("el significado semántico de las oraciones por encima de las palabras sueltas.")
```
