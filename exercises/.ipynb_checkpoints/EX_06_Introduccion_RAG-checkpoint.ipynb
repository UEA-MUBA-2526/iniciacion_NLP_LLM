## Actividad 1 — Plantilla de contexto

Escribe una función `build_prompt(context_chunks, question) -> str` que inserte los pasajes en un delimitador claro (`### Context` / `### Question`).


```python
def build_prompt(context_chunks: list[str], question: str) -> str:
    joined_context = "\n\n".join([f"- {chunk.strip()}" for chunk in context_chunks])
    
    prompt = f"""Eres un asistente de IA experto que responde preguntas basándose única y exclusivamente en el contexto provisto a continuación. 
Si la respuesta no se encuentra en el contexto, abstente de responder y di de forma clara: "No dispongo de información suficiente".

### Context:
{joined_context}

### Question:
{question}

### Answer:
"""
    return prompt


chunks_ejemplo = [
    "El protocolo HTTP corre en el puerto 80 por defecto.",
    "El protocolo seguro HTTPS utiliza cifrado TLS/SSL y corre en el puerto 443."
]
pregunta_ejemplo = "¿Qué puerto utiliza HTTPS?"

prompt_confeccionado = build_prompt(chunks_ejemplo, pregunta_ejemplo)
print("=== ACTIVIDAD 1: PROMPT DE CONTEXTO GENERADO ===")
print(prompt_confeccionado)
```

## Actividad 2 — RAG sin LLM (retrieval only)

Con tus chunks del notebook teórico (o texto inventado), recupera top-k y **imprime** el contexto ensamblado sin llamar al generador.


```python
# TODO: retrieve top-k for a fixed question

from sentence_transformers import SentenceTransformer
import numpy as np

model_rag = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device="cpu")

# Base de datos de conocimiento indexada
chunks_base_conocimiento = [
    "Los árboles frutales como el manzano requieren una poda sistemática a finales del invierno.",
    "La arquitectura de los Transformers revolucionó el NLP gracias al mecanismo de auto-atención (self-attention).",
    "Los modelos RAG combinan una etapa de recuperación de información indexada con un modelo generativo final.",
    "El cultivo hidropónico permite hacer crecer plantas utilizando soluciones minerales en lugar de suelo agrícola."
]

# Generar embeddings para toda nuestra base de conocimiento
embeddings_chunks = model_rag.encode(chunks_base_conocimiento)

# Definición de la pregunta fija (Fixed question) y codificación de su vector
fixed_question = "¿Cómo funciona la estructura y metodología de un sistema RAG?"
query_embedding = model_rag.encode(fixed_question)

# TODO: retrieve top-k for a fixed question (Calculamos similitud coseno nativa)
def cos_sim(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

scores = [cos_sim(query_embedding, chunk_emb) for chunk_emb in embeddings_chunks]

# Extraemos los top-2 fragmentos con mayor score
top_k = 2
top_k_indices = np.argsort(scores)[::-1][:top_k]
contextos_recuperados = [chunks_base_conocimiento[idx] for idx in top_k_indices]

prompt_final_rag = build_prompt(contextos_recuperados, fixed_question)

print("=== ACTIVIDAD 2: CONTEXTO RAG RECUPERADO Y ENSAMBLADO ===")
print(prompt_final_rag)
```
