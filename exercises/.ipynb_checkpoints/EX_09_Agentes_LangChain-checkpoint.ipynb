## Actividad 1 — Tool decorator

Define un `@tool` (LangChain) que calcule el número de palabras de un texto. Prueba `.invoke` con un string.

*Hint:* `from langchain_core.tools import tool`.


```python
from langchain_core.tools import tool

@tool
def count_words(text: str) -> int:
    """Calcula el número de palabras que contiene un texto de entrada."""
    return len(text.split())

# Probamos la ejecución directa de la herramienta con .invoke
resultado = count_words.invoke("I love RAG")
print(f"Resultado de invoke: {resultado}") 
```
